## Cell 3 — EDGAR parser verification. **STOP HERE AND READ.**

5 seconds, 4 API calls. This is the cheapest bug catch in the project.

| Symbol | Expected timing | Why |
|---|---|---|
| AAPL | `amc` | reports after the close |
| MSFT | `amc` | reports after the close |
| JPM  | `bmo` | banks report before the open |

- **All three identical** → parser broken. Do not continue.
- **AAPL `intraday` year-round** → set `SEC_TS_IS_UTC = False` in Cell 3b, rerun.
- **AAPL `intraday` in summer, `amc` in winter** → timezone handling failed; check `zoneinfo`.
- **amc / amc / bmo** → correct. Continue.


```python
# CELL 3 - VERIFY. STOP AND READ THE OUTPUT.
!python build_earnings.py --symbols AAPL,MSFT,JPM --years 4 --out _verify.csv

import csv, collections
rows = list(csv.DictReader(open("_verify.csv")))
by = collections.defaultdict(list)
for r in rows: by[r["symbol"]].append(r)
print()
for s in ("AAPL","MSFT","JPM"):
    v = by.get(s, [])
    tim = collections.Counter(x["timing"] for x in v)
    mon = sorted({x["date"][5:7] for x in v})
    print(f"{s:6} n={len(v):3}  timing={dict(tim)}  months={mon}")
print()
ok = (by and all(by.get(s) for s in ("AAPL","MSFT","JPM"))
      and collections.Counter(x["timing"] for x in by["AAPL"]).most_common(1)[0][0]=="amc"
      and collections.Counter(x["timing"] for x in by["JPM"]).most_common(1)[0][0]=="bmo")
print("VERIFICATION:", "PASS - continue" if ok else "FAIL - do not continue, see Cell 3b")
```
